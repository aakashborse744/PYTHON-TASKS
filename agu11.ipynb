# Assignment operator (=):-> Used to assign a value to a variable.
# a = 10
# value assign to variable 10

# Composite Assignment:-> These combine an arithmetic/bitwise operation with assignment in one step. x = x op value becomes x op= value.
#  += , -= , *= , /= , //= , %= , **= .

```python
a = 10
b = a + 5  #value assign to var 'a' is 10 and value assign to 'b' is 'a+5' so the  value 'a' is 10 so in print it will show 10+5=15 
print(b)
a += 5
print(a)

a = 10
a = a + 5
print(a)      #value store only one variable 'a'
a += 5
print(a)
```

```python
a = 8
a += 12        
print("a =",a)     #value store only one variable 'a'

b = 5
b -= 3
print(" b =",b)     #value store only one variable 'b'

c = 9
c *= 3
print(" c =",c)

d = 12
d /= 5
print(" d =",d)

e = 12
e //= 5
print(" e =",e)

f = 14
f %= 5
print(" f =",f)

g = 8
g **= 6
print(" g =",g) 


```

```python
# Decimal:-> decimal is base-10 — it only uses digits 0 to 9.

a = 8
print(a)
print(type (a))
```

```python
# Binary:-> binary is base-2 — it only uses digits 0 to 1.
var = 0b1010
var2 = 0B1010
print(var)
print(type(var))
```

```python
# octal :-> Octal is base-8 — it only uses digits 0 to 7 (no 8 or 9).
var = 0o1010
# var = 0o678
print(var)
```

```python
# Hexadecimal:->Octal is base-8 — it only uses digits 0 to 9 , a to f and A to F(no g to z).
var = 0xface
print(var)
```

# BItwise operator
# & , | , ^ , >> , << , >>>

```python
a = 5               #  0101 
b = 3               #  0011
print(a & b)     #-->  0001  #1
                
```

```python
a = 5           # 0101
b = 3           # 0011
print(a | b) #--> 0111   #7
```

```python
a = 3          # 0011
b = 7          # 0111
print(a ^ b) #-->0100  #4
```

```python
# Right shift (>>):->any no. shift right side so no. is decreasing.
a = 5            # 000 0101              # Right shift Formula
b = 2       #shift     0001              # 5 / 2 ^ 2
print(a >> b)                            # 5 // 4   =1

a = 8            # 000 1000              # 8 / 2 ^ 2
b = 2       #shift     0001              # 8 // 4   =2
print(a >> b)
```

```python
# Left shift:->any no. shif left side so no. is increasing.
a = 5               # 000 0101             # left shift Formula
b = 2        #shift    01 0100             # 5*2^2
print(a << b)                              # 5*4  =20
                                             

a = 8          # 000 1000                  # 8 * 2 ^ 2 
b = 2      #shift 10 0000                  # 8 * 4   =32
print(8 << b)                                         
                                                                            
```
