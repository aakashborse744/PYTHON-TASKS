# 5 slicing operations on my name.
# negatitive index


```python
s1 = "Aakash"
print(s1[-6 : -1])   # Aakas
print(s1[-4 : -1])   # kas
print(s1[-5 : -1])   # akash
print(s1[-6 : -2])   # Aaka
print(s1[-6 : -5])   # A

print(s1[-4:])  # 'ash' -> last 3 characters
print(s1[::-4])
```

# combination of negative as well as positive

```python
s1 = "Aakash"
print(s1[-6 : 6])   # Aakash
print(s1[0 : -1])   # Aakas   # internally -1
print(s1[-6 : 5])   # Aakas
print(s1[-4 : 6])   # kash 
print(s1[2 : -1])   # kas      # can't add any value in -1.
```
