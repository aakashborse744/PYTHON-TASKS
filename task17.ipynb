```python
file = "myfile12345.pdf"
count = file.index(".")
name = file[0:count]
print(name.upper())          #MYFILE12345
print(name.title())          #Myfile12345
print(name.isalnum())        #true

extention=file[count+1:]
print(extention)     #PDF
print(extention.isalnum())   #true
print(extention.isdigit())   #false

```

```python
file = "presentation.pptx"
count=file.index(".")
name=file[0:count]
print(name.split())     #list of string :-['presentation']
print(name.upper())     #string :-PRESENTATION
print(name.isalnum())   #bool :-true
print(name.title())     #string :-presentation
print(file.find("."))   #integer :-12

extention=file[count+1:]
print(extention.lower())       #string :-pptx
print(extention.swapcase())    #string :-PPTX
print(extention.isdigit())     #bool :-false
print(file.endswith(".pptx"))  #bool :-true
print(file.count("p"))         #integer — kitni baar wo character aaya → 3



```
