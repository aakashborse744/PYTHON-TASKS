## 1>append:-Add single item end of the list.

```python
student_db = []
student_db.append('aakash')
student_db.append('tushar')
student_db.append('rohit')
print(student_db)
print(id(student_db))
student_db.append('nilesh')
student_db.append('mayur')
student_db.append('mayur')
print(student_db)
print(id(student_db))
```

## 2>extend(iterable):-Add all item from another iterable to the end.

```python
print(student_db)
student_db.extend(['ganesh'])
student_db.extend(['golu'])
print(student_db)
```

## 3>insert:-Insert item x at index student_db.

```python
print(student_db)
student_db.insert(3 ,5)
print(student_db)
```

## 4>remove:-Remove the first matching value x.

```python
print(student_db)
name = input("Enter a name =")
if name in student_db:
    student_db.remove(name)
    print(" removed successfully")
else:
    print(' already remove list')
print(name)
print(student_db)
```

## 5>pop:-remove and return item at index ,student_db (default:last item).

```python
print(student_db)
student_db.pop(4)
print(student_db)
```

## 6>clear:-remove all items from the list.

```python
print(student_db)
student_db.clear()
print(student_db)
```

## 7>index:-return the index of the first match x.
# show position in the index.

```python
print(student_db)
pos = student_db.index('mayur')
print(pos)
```

## 8>count:-count how many times x appears.

```python
print(student_db)
count = student_db.count("mayur")
print(count)
```

## 9>sort():-sort the list in place (accending by default).

```python
print(student_db)
student_db.sort()
```

## 10>reverse():-Reverse the list in place. (descending by default)

```python
print(student_db)
student_db.reverse()
```

## 11>copy:-Return shallow copy(same list created in another list) of the list.

```python
print(student_db)
new_db = student_db.copy()
print(new_db)
```
