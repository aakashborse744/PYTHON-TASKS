# Task 2 :->Write a program to determine if a person is a minor, adult, or senior citizen based on age.

```python

age= int(input("Enter age:"))

if age < 25:
    print(age,"minor age")
elif age < 50:
    print(age,"adult")
else:
    print(age,"senior citizen")
```
