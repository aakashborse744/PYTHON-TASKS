# Conditional Statements

# if :-
#    To check half conditon.


# if-else 
#      we use if-else to check one complete condition
#      .only one condition is work at one time, if-else is not will  work one time.
#                .condition is true so work if.
#                .condition is false so automatically work else.
# syntax:-  if condition :
#                if block code 

#           else :
#                else block code     

```python
tushar = 23
akash =23
#who will pay the bill 
if tushar < akash: #false     
    print("tushar will pay the bill")

elif tushar == akash:
    print("tushar , akash both will pay the bill")    

else :
    print("akash will pay the bill")
print("End the program")       
```
